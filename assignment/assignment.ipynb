```python
# client is your MongoDB connection from the previous code.
# sample_mflix is the database name.
# db now represents the sample_mflix database.
# f indicates python includes a variable in print statement 
# count_documents() gives us the number of documents (movies) inside the movies collection.
db = client.sample_mflix
movies = db.movies
print(f"📊 Database: {db.name}")
print(f"📁 Collection: movies ({movies.count_documents({})} documents)")
```


### Question 1

Question: From the `movies` collection, return the documents with the `plot` that starts with `"war"` in acending order of released date, print only title, plot and released fields. Limit the result to 5.


**Answer:**

```python
# ============================================================================
# Question 1
# Question: From the `movies` collection, return the documents with the `plot` that starts with `"war"` in acending order of released date, print only title, plot and released fields. Limit the result to 5.
# ============================================================================
query = {
    "plot": {
        "$regex": "war",
        "$options": "i"
    }
}

projection = {
    "_id": 0,
    "title": 1,
    "plot": 1,
    "released": 1
}

for movie in movies.find(query, projection).sort("released", 1).limit(5):
    print(f"Title: {movie.get('title')}")
    print(f"Released: {movie.get('released')}")
    print(f"Plot: {movie.get('plot')}")
```

### Question 2

Question: Group by `rated` and count the number of movies in each.

**Answer:**

```python
# ============================================================================
# Question 
# Question: Group by `rated` and count the number of movies in each.
# ============================================================================

pipeline = [
    {
        "$group": {
            "_id": "$rated",
            "count": {"$sum": 1}
        }
    },
    {
        "$sort": {
            "count": -1
        }
    }
]

for r in movies.aggregate(pipeline):
    print(r)
```


### Question 3

Question: Count the number of movies with 3 comments or more.


**Answer:**


```python
# ============================================================================
# Question 3
# Question: Count the number of movies with 3 comments or more.
# ============================================================================

# Look at the comments collection.
# Group comments by movie_id.
# Count comments for each movie.
# Keep movies having at least 3 comments.
# Count how many such movies there are.

comments = db.comments

# Calculate the number of comments for each movie:
stage_groupby_movieid = {
    "$group": {
                "_id": "$movie_id",
                "count": {"$sum": 1}
            }
}

stage_match_with_comments = {
    "$match": {
        "count": {"$gte": 3}
    }
}

stage_count_movies_with_3overcomments = {
        "$count": "movies_with_3_or_more_comments"
}

pipeline = [
    stage_groupby_movieid,
    stage_match_with_comments,
    stage_count_movies_with_3overcomments
]


results = comments.aggregate(pipeline)

for comments in results:
    print(comments)
    
```
